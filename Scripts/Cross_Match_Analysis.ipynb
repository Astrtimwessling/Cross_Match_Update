# Updated Match Diagnostics

```python
cross_match_updated = Table.read(output_dir + 'cross_match_updated.fits')
cross_match_updated.pprint_all()
```

```python
updated_fermi_names = cross_match_updated['fermi_name_counterpart'].tolist()
updated_swift_names = cross_match_updated['bat_name_counterpart'].tolist()

updated_matches = list(zip(updated_swift_names, updated_fermi_names))
```

```python
updated_swift_types = cross_match_updated['bat_category_type'].tolist()
updated_fermi_types = cross_match_updated['fermi_category_type'].tolist()
print(Counter(updated_swift_types))
print(Counter(updated_fermi_types))

updated_match_types = list(zip(updated_swift_types, updated_fermi_types))
print(Counter(updated_match_types))
```

# Comparison to first Cross-Match

```python
first_cross_match = Table.read(catalog_dir + 'crossmatch_SED_latest.fits')
print(first_cross_match.colnames)
```

```python
first_match_fermi_names = [first_cross_match['fermi_name_counterpart'].tolist()[x] if first_cross_match['fermi_name_counterpart'].tolist()[x] != '' else first_cross_match['fermi_name_counterpart2'].tolist()[x] for x in range(len(first_cross_match))]
first_match_swift_names = first_cross_match['bat_name_counterpart'].tolist()

first_matches = list(zip(first_match_swift_names, first_match_fermi_names))
```

## Retained Matches

```python
retained_matches = [x for x in updated_matches if x in first_matches]
new_matches = [x for x in updated_matches if x not in first_matches]
not_rematched = [x for x in first_matches if x not in updated_matches]

print(f"Retained matches: {len(retained_matches)}")
print(f"New matches found in cross-match update: {len(new_matches)}")
print(f"Matches in first match but not in update: {len(not_rematched)}")
```

```python
new_swift_counterparts = [x for x in updated_swift_names if x not in first_match_swift_names]
not_rematched_swift_counterparts = [x for x in first_match_swift_names if x not in updated_swift_names]

print(f"New swift counterparts found in cross-match update: {len(new_swift_counterparts)}")
print(f"Swift counterparts in first match but not in update: {len(not_rematched_swift_counterparts)}")

for i in range(len(not_rematched_swift_counterparts)):
    print(not_rematched_swift_counterparts[i])
```

```python
new_fermi_counterparts = [x for x in updated_fermi_names if x not in first_match_fermi_names]
not_rematched_fermi_counterparts = [x for x in first_match_fermi_names if x not in updated_fermi_names]

print(f"New fermi counterparts found in cross-match update: {len(new_fermi_counterparts)}")
print(f"fermi counterparts in first match but not in update: {len(not_rematched_fermi_counterparts)}")

for i in range(len(not_rematched_fermi_counterparts)):
    print(not_rematched_fermi_counterparts[i])
```

```python
for x in range(len(not_rematched)):
    if not_rematched[x][0] not in not_rematched_swift_counterparts:
        for i in range(len(new_matches)):
            if not_rematched[x][0] == new_matches[i][0]:
                print(not_rematched[x], new_matches[i])
    if not_rematched[x][1] not in not_rematched_fermi_counterparts:
        for i in range(len(new_matches)):
            if not_rematched[x][1] == new_matches[i][1]:
                print(not_rematched[x], new_matches[i])
                
# Matches to add to retained:
Additional_Retained_Matches = [('SWIFT_J1254.9+1165', 'ON 187'),  ('SGR A*', 'Galactic Center'), ('Crab', 'Crab Nebula')]
Additional_Retained_Matches_Old_Names = [('SWIFT J1254.9+1165', 'ON 187'), ('SGR A*', 'Galactic Centre'), ('Crab', 'Crab Nebula$^\\dagger$')]
```

```python
updated_separation = cross_match_updated['separation'].tolist()
first_match_separation = first_cross_match['separation'].tolist()

updated_flags = cross_match_updated['flag'].tolist()
first_match_flags = first_cross_match['Flag'].tolist()

first_match_swift_types = first_cross_match['bat_category_type'].tolist()
first_match_fermi_types = first_cross_match['fermi_category_type'].tolist()
first_match_types = list(zip(first_match_swift_types, first_match_fermi_types))

delta_separation = []
old_flag = []
old_types = []

for x in range(len(updated_separation)):
    if updated_matches[x] in retained_matches:
        idx = first_matches.index(updated_matches[x])
        delta_sep = updated_separation[x] - first_match_separation[idx]
        delta_separation.append(delta_sep)
        
        old_flag.append(first_match_flags[idx])
        old_types.append(first_match_types[idx])
        
    elif updated_matches[x] in Additional_Retained_Matches:
        idx_additional = Additional_Retained_Matches.index(updated_matches[x])
        old_match_name = Additional_Retained_Matches_Old_Names[idx_additional]
        
        idx = first_matches.index(old_match_name)
        delta_sep = updated_separation[x] - first_match_separation[idx]
        delta_separation.append(delta_sep)
        
        old_flag.append(first_match_flags[idx])
        old_types.append(first_match_types[idx])
    
    else:
        delta_separation.append('NA')
        old_flag.append('NA')
        old_types.append('NA')
        
print(delta_separation)
```

```python
cross_match_updated['delta_sep'] = delta_separation
cross_match_updated['first_match_flag'] = old_flag

cross_match_updated.pprint_all()
```

```python
for i in range(len(updated_flags)):
    if old_flag[i] != 'NA' and old_flag[i] != updated_flags[i]:
        print(f"Old Flag = {old_flag[i]}, New Flag = {updated_flags[i]}, Old Types = {old_types[i]}, New Types = {updated_match_types[i]}, {updated_matches[i]}")
```
